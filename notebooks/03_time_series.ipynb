```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
df = pd.read_csv("C:/MADS/DATEXP03/Smart Toilet/data/raw/dataset 1.csv")

df.head(1)
```

```python
# New features for time series analysis (time in seconds accounting for window, relative time and session duration) and sorting the data by session and window
df = df.sort_values(["session", "window"])
df["session_duration"] = df.groupby("session")["window"].transform("max") + 1
df["relative_time"] = df["window"] / df.groupby("session")["window"].transform("max")
df["time_seconds"] = df["window"] * 2
```

```python
features = [
    "relative_time",
    "session_duration",
    "time_seconds",
    "weight",
    "heartrate",
    "temperature",
    "comx",
    "comy"
]
```

```python
X = []
y = []

for session, g in df.groupby("session"):
    
    seq = g[features].values
    X.append(seq)

    y.append(g["user_id"].iloc[0])   # of event label
```

```python
X = pad_sequences(X, padding="post", dtype="float32")
```
